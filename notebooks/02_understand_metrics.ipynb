# 02 - Understand the metrics

doc-bench scores parser output with six deterministic metrics, all in
`[0, 1]` (1.0 = perfect): **NID, TEDS, MHS, ARD, BLEU, METEOR**. They compare
*gold markdown* against *prediction markdown*. Here we compute them on toy
inputs to build intuition. See `docs/doc-bench/metrics.md` for definitions.

```python
from doc_bench.metrics.parsing.nid import nid_score, nid_s_score
from doc_bench.metrics.parsing.table_teds import teds_score, teds_s_score
from doc_bench.metrics.parsing.mhs import mhs_score, mhs_s_score
from doc_bench.metrics.parsing.reading_order import ard_score
from doc_bench.metrics.parsing.text_similarity import bleu_score, meteor_score
```

## A gold document and two predictions

```python
GOLD = '''# Quarterly Report

## Revenue

Revenue grew across all regions this quarter.
'''

PRED_GOOD = '''# Quarterly Report

## Revenue

Revenue grew across all regions this quarter.
'''

PRED_POOR = 'Quarterly Report Revenue Revenue grew across regions.'
```

## Score each prediction

```python
def score(gold, pred):
    return {
        'NID': round(nid_score(gold, pred), 4),
        'MHS': round(mhs_score(gold, pred), 4),
        'BLEU': round(bleu_score(gold, pred), 4),
        'METEOR': round(meteor_score(gold, pred), 4),
        'ARD': round(ard_score(pred.split(), gold.split()), 4),
    }

print('good:', score(GOLD, PRED_GOOD))
print('poor:', score(GOLD, PRED_POOR))
```

Notice how the structure-preserving prediction scores near 1.0 on NID/MHS,
while the flattened one collapses on MHS (no headings) and BLEU.

**Caveats** (see `docs/doc-bench/metrics.md`):
- TEDS/MHS need markdown tables/headings in the gold to be meaningful.
- METEOR needs NLTK data (`doc-bench-setup`) or it reads 0.
- Align the prediction's page coverage to the gold, or text metrics deflate.
