# 01 - Explore the bundled fixtures

doc-bench ships **33 documents** inside the wheel so you can evaluate without
downloading anything. This notebook reads the packaged `manifest.json`, shows
the dataset composition, and peeks at one ground-truth file.

> Prereq: `doc-bench` is importable (run `uv sync`, or `pip install` the wheel).

```python
import json
from collections import Counter
from importlib.resources import files

fixtures = files('doc_bench') / 'fixtures'
manifest = json.loads((fixtures / 'manifest.json').read_text())
print(manifest['name'], '-- total docs:', manifest['total'])
```

## Composition by dataset

```python
for dataset in ('dp_bench', 'omnidocbench', 'ato_bench'):
    entries = manifest.get(dataset, [])
    key = 'category' if dataset == 'dp_bench' else 'doc_type'
    counts = Counter(e.get(key, 'unknown') for e in entries)
    print(f'{dataset}: {len(entries)} docs -> {dict(counts)}')
```

## Peek at a ground-truth annotation

The bundled ATO document uses the OmniDocBench `layout_dets` format (one file
per page). Each detection carries a `category_type`, a `poly` box, an `order`,
and `text`.

```python
ato = manifest['ato_bench'][0]
print('doc_id:', ato['doc_id'], '| pages:', ato['pages'])
page1 = json.loads((fixtures / ato['pages'][0]).read_text())
dets = page1['layout_dets']
print('page 1 detections:', len(dets))
print('categories:', Counter(d['category_type'] for d in dets))
print('first text det:', next(d['text'] for d in dets if d.get('text')))
```
